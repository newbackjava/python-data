```python
import nltk
from konlpy.corpus import kobill
```

```python
files_ko = kobill.fileids()
doc_ko = kobill.open('1809890.txt').read()
```

```python
doc_ko
```

```python
from konlpy.tag import Okt
```

```python
t = Okt()
```

```python
tokens_ko = t.nouns(doc_ko)
# tokens_ko
```

```python
## 불용어 처리할 부분
#
#
#
#
```

```python
# nltk를 이용해서 빈도수를 구하고 그래프로 그려보자
# Text()로 만들고, 그래프로 그리고! 순서
```

```python
ko = nltk.Text(tokens_ko)
```

```python
plt.figure(figsize=(16, 8))
ko.plot(10)
plt.show()
```

```python
from wordcloud import WordCloud
```

```python
data = ko.vocab().most_common(100) #워드 클라우드에 10위까지 넣음.
```

```python
type(data)
```

```python
path = '/usr/share/fonts/truetype/nanum/NanumBarunGothic.ttf'
wc = WordCloud(font_path=path, relative_scaling=0.2,
               background_color='white').generate_from_frequencies(dict(data))
```

```python
plt.figure()
plt.imshow(wc) ##### 워드클라우드 객체
plt.axis('off') #x,y축제거
plt.show()
```

```python
# 불용어 리스트 (원하는 대로 계속 추가 가능)
stopwords = [
    '있습니다', '합니다', '합니다.', '수', '것', '등', '및', '통해', '대한',
    '이', '그', '저', '그리고', '또한', '그래서', '때문', '때문에',
    '이번', '이번에', '오늘', '지난', '관련', '제공', '등의',
    '으로', '에서', '에게', '까지', '부터',
    '차', '위', '터', '대의', '이번', '김건희', '윤석열', '제', '모든', '특검'
    # 필요하면 여기 계속 추가하면 됨!
]
```

```python
tokens_ko = [ x for x in tokens_ko if len(x) >= 1 and x not in stopwords]
tokens_ko[:10]
```

```python
ko = nltk.Text(tokens_ko, name='test')
```

```python
plt.figure(figsize=(16,8))
ko.plot(10)
plt.show()
```
