```python
test_df['document'] = test_df['document'].apply(lambda x: re.sub(r'[^ ㄱ-ㅣ가-힣]+', "", x))
test_df
```

```python
# 사전을 만들어야한다.
# 자연어를 파이썬에서 인식할 때는 단어별 번호로 인식함.
# 나는 별로 였어라고 했다고 한다면 나는(1), 별로(100) --> 부정사전, 부정이라고 판단
```

```python
# !pip install konlpy
```

```python
from konlpy.tag import Okt

okt = Okt()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
## 사전을 만들어주고, df의 문장을 원핫인코딩으로 컬럼을 변환시켜 전환해줌.
# 각 단어에 대해 번호를 부여함.(빈도수 + 이상하게자주반복되는여부)
tfidf = TfidfVectorizer(tokenizer= okt.morphs , ngram_range=(1, 2), min_df = 3, max_df = 0.9)
tfidf
```

```python
tfidf.fit(train_df['document']) #사전만들어라.
```

```python
train_tfidf_df = tfidf.transform(train_df['document'])
train_tfidf_df #사전을 기준으로 어떤 문장들이 왔는지 원핫인코딩으로 컬럼을 증가시켜 표현해라
```

```python
train_tfidf_df[0].toarray()
```

```python
train_tfidf_df[:2].toarray()
```

```python
tfidf.vocabulary_
```

```python
tfidf.inverse_transform(train_tfidf_df[0])
```

```python
from sklearn.linear_model import LogisticRegression

SA_lr = LogisticRegression(random_state = 0)
```

```python
SA_lr.fit(train_tfidf_df, train_df['label'])
```

```python

#st = input('감성 분석할 문장입력 >> ')
st = "웃자 ^o^ 오늘은 좋은 날이 될 것 같은 예감100%! ^^*"
```

```python

st2 = re.sub(r'[^ ㄱ-ㅎ|가-힣]+', "", st)
st2
```

```python

st_tfidf_df = tfidf.transform([st2])
st_tfidf_df
```

```python

st2_predict = SA_lr.predict(st_tfidf_df)
```

```python

if(st2_predict== 0):
    print(st2 , "->> 부정 감성")
else :
    print(st2 , "->> 긍정 감성")
```
