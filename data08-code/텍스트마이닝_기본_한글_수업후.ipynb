```python
import json #json형식의 파일을 dict변환시켜줌.
```

```python
file_name = '/content/etnews.json'
json_data = json.loads(open(file_name, 'r', encoding='utf-8').read())
json_data
```

```python
len(json_data),type(json_data)
```

```python
all_data = ""
```

```python
import re #정규식(regular expression)
#문자열 패턴을 지정하여 필터링가능
#전화번호 숫자3-숫자4-숫자4
#우편번호 숫자6-숫자7
```

```python
for item in json_data : #dict의 리스트
  # print(item['message'])
  # re.sub('정규식', '', 대상)
  all_data += re.sub(r'[^\w]' , ' ', item['message']) + " "
```

```python
print(all_data)
```

```python
noun_list3 = okt.nouns(all_data)
noun_list3[:3]
```

```python
count3 = Counter(noun_list3)
```

```python
count3.most_common(5)
```

```python
most10 = dict(count3.most_common(10))
most10
```

```python
plt.figure(figsize=(12,5))
plt.xlabel('키워드')
plt.ylabel('빈도수')
plt.grid(True)

plt.bar(range(len(most10)), most10.values())
plt.xticks(range(len(most10)), most10.keys())
plt.show()
```

```python
# 불용어 리스트 (원하는 대로 계속 추가 가능)
stopwords = [
    '있습니다', '합니다', '합니다.', '수', '것', '등', '및', '통해', '대한',
    '이', '그', '저', '그리고', '또한', '그래서', '때문', '때문에',
    '이번', '이번에', '오늘', '지난', '관련', '제공', '등의',
    '으로', '에서', '에게', '까지', '부터',
    '차', '위', '터', '대의', '이번'
    # 필요하면 여기 계속 추가하면 됨!
]
```

```python
# stopwords = set(stopwords)
```

```python
stop_result = [ x for x in noun_list3 if len(x) >= 1 and x not in stopwords]
stop_result[:10]
```

```python
# 카운트를 해서 수정된 내용으로 차트그려보세요.
# stop_result로 카운트 처리
# 상위 10등까지 추출
# 그래프를 그리기 위해서 dict()로 타입변환
# 그래프 그리기
```

```python
count4 = Counter(stop_result)
count4.most_common(10)
```

```python
stop10 = dict(count4.most_common(10))
stop10
```

```python
plt.figure(figsize=(12,5))
plt.xlabel('키워드')
plt.ylabel('빈도수')
plt.grid(True)

plt.bar(range(len(stop10)), stop10.values())
plt.xticks(range(len(stop10)), stop10.keys())
plt.show()
```
