```python
import requests
import xmltodict
import json


def main():
    # 1. API URL 설정
    url = (
        "https://www.youthcenter.go.kr/opi/empList.do"
        "?openApiVlak=5c0e90c4dbb47a74b3a37e0b"  # 예시 키, 실제 키로 교체
        "&pageIndex=1"
        "&display=100"
    )

    # 2. API 호출 (XML 문자열 받아오기)
    response = requests.get(url)
    response.raise_for_status()  # 오류 발생 시 예외

    xml_text = response.text

    # 3. XML → 파이썬 dict (JSON 객체 비슷)
    jobj = xmltodict.parse(xml_text)

    # 변환된 전체 구조 확인 (Java: jobj.toString())
    print("===== 전체 JSON 구조 =====")
    print(json.dumps(jobj, ensure_ascii=False, indent=2))

    # 4. 필요한 노드까지 내려가기
    #    (실제 태그 이름은 응답 XML 구조에 맞게 수정)

    # 예시: <response><empsInfo><emp>...</emp></empsInfo></response> 라고 가정
    response_node = jobj.get("response", {})          # 1단계
    emps_info = response_node.get("empsInfo", {})     # 2단계 (Java 의 jobj1 역할)

    # 5. 배열(리스트) 꺼내기
    #    예: <emp> 태그들이 배열 형태라고 가정
    emp_list = emps_info.get("emp", [])

    # emp 가 1개일 때는 dict 하나로 올 수 있으므로 리스트로 변환
    if isinstance(emp_list, dict):
        emp_list = [emp_list]

    print("===== 직원 리스트 =====")
    for emp in emp_list:
        # XML 구조에 맞는 필드 이름 사용 (예: name, id, empNm 등)
        name = emp.get("name")
        emp_id = emp.get("id")
        # 필요하면 더 많은 필드 꺼내기
        print(f"이름: {name}, ID: {emp_id}")


if __name__ == "__main__":
    main()
```
