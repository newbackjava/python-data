```python
import requests
```

```python
# 1. 기본 URL
url = "http://api.koreafilm.or.kr/openapi-data2/wisenut/search_api/search_json2.jsp"

# 2. 요청 파라미터
params = {
    "collection": "kmdb_new2",
    "nation": "대한민국",
    "ServiceKey": "R71XD4VMIF4UV7I452XI",
    "val001": "2018",  # 상영년도
    "val002": "01"     # 상영월
}

# 3. GET 요청
response = requests.get(url, params=params)
response
```

```python
# 4. JSON으로 변환
data = response.json()

# 5. 영화 리스트 위치까지 들어가기
#   최상위: data
#   -> 'Data' (리스트)
#   -> Data[0]
#   -> 'Result' (여러 영화가 들어 있는 리스트)
movies = data["Data"][0]["Result"]
movies
```

```python
# 6. 여러 영화 정보 꺼내기
for idx, movie in enumerate(movies):
    # title = movie.get("title")           # 제목
    title =movie['title']

    prod_year = movie.get("prodYear")    # 제작/개봉년
    nation = movie.get("nation")         # 국가

    # 감독 정보는 directors -> director(리스트) 안에 있음
    directors_total = movie.get("directors")
    print("directors_total---- ", directors_total) #dic

    directors_info = directors_total.get("director") #list
    print("directors_info---- ",directors_info)

    director_names = ""
    for one in directors_info:
        print("one---- ",one)
        director_names += one.get("directorNm") + " "

    print("director_names ---- ", director_names)

    print(f"    제목: {title}")
    print(f"    제작년도: {prod_year}")
    print(f"    국가: {nation}")
    print(f"    감독: {director_names}")
    print("-" * 40)
```
