```python
import os
import sys
import json
import urllib.request
import urllib.parse
import time   # (선택) 너무 빠른 연속 호출을 피하고 싶을 때 사용

client_id = 'hfhBxkqxdEy5alw1C5Rs'   # 실제 사용 시 본인 키로 교체 권장
client_secret = 'rHJC0Vf0W5'

# 검색어
encText = urllib.parse.quote("비트코인")

# 한 페이지당 결과 수 (기본 10)
DISPLAY = 10

# 가져올 페이지 수
TOTAL_PAGES = 20   # <-- 여기만 바꿔서 페이지 수 조절

all_items = []   # 전체 결과를 담을 리스트

for page in range(1, 21):
    # start 값 계산: 1, 11, 21, ... 이런 식으로 증가
    start = (page - 1) * DISPLAY + 1

    url = (
        "https://openapi.naver.com/v1/search/blog"
        f"?query={encText}&display={DISPLAY}&start={start}"
    )

    request = urllib.request.Request(url)
    request.add_header("X-Naver-Client-Id", client_id)
    request.add_header("X-Naver-Client-Secret", client_secret)

    response = urllib.request.urlopen(request)
    rescode = response.getcode()

    if rescode == 200:
        response_body = response.read()
        res_data = response_body.decode('utf-8')

        json_data = json.loads(res_data)    # dict

        # 이번 페이지의 item만 따로 추출
        items = json_data.get("items", [])
        all_items.extend(items)

        print(f"{page} 페이지 수신 완료 (start={start}, items={len(items)})")
        # 너무 빨리 여러 번 호출하는 게 걱정되면 잠깐 쉬기
        time.sleep(0.1)

    else:
        print("Error Code:" + str(rescode))
        break

print("\n총 수집 개수:", len(all_items))

# 예시: 앞의 5개 제목만 출력
for i, item in enumerate(all_items[:5], start=1):
    print(f"[{i}] {item.get('title')}")
```
